```python
import pandas as pd

# Đọc dữ liệu Wine
df = pd.read_excel("../D00 - wine.data.xlsx", sheet_name="wine.data")

# Khám phá dữ liệu
print("Kích thước:", df.shape)
display(df.head())
print("Tên các cột:", df.columns.tolist())
```

```python

# Kiểm tra thông tin dữ liệu
df.info()

# Kiểm tra giá trị thiếu
print("\nGiá trị thiếu:")
print(df.isnull().sum())

# Kiểm tra dữ liệu trùng lặp
print("\nSố dòng trùng lặp:", df.duplicated().sum())

# Thống kê mô tả
display(df.describe())
```

Nhận xét: Bộ dữ liệu Wine gồm 178 mẫu và 14 cột. Các cột đều có kiểu dữ liệu số, không phát hiện giá trị thiếu hoặc dòng trùng lặp. Vì vậy, dữ liệu không cần xử lý thiếu hoặc trùng lặp trước khi tiếp tục phân tích.

```python
# Tạo bản sao và bỏ cột cuối
df_clean = df.iloc[:, :-1].copy()

# Tách thuộc tính và nhãn
X = df_clean.drop(columns=["Class"])
y = df_clean["Class"]

print("Kích thước sau xử lý:", df_clean.shape)
print("Số thuộc tính:", X.shape[1])
print("Số lớp:", y.nunique())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Biểu đồ phân bố 3 lớp rượu
sns.countplot(data=df_clean, x="Class")
plt.title("Wine Class Distribution")
plt.show()

# Heatmap tương quan giữa các thuộc tính
plt.figure(figsize=(10, 8))
sns.heatmap(X.corr(), cmap="coolwarm")
plt.title("Feature Correlation")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Chuẩn hóa 12 thuộc tính
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Giảm dữ liệu xuống 2 chiều
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Vẽ biểu đồ PCA
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=y,
    palette="Set1"
)


plt.title("PCA 2D - Wine Dataset")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Chuẩn hóa 12 thuộc tính
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Giảm dữ liệu xuống 2 chiều
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Vẽ biểu đồ PCA
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=y,
    palette="Set1"
)

plt.title("PCA 2D - Wine Dataset")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```
