```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV, KFold, cross_val_score
from sklearn.naive_bayes import GaussianNB

# 1. Tải và chuẩn hóa dữ liệu Wine
wine = load_wine()
X, y = wine.data, wine.target
scaler = StandardScaler()
X = scaler.fit_transform(X)

# 2. PHẦN VIỆC CỦA VÂN: GaussianNB & GridSearchCV
gnb = GaussianNB()
param_grid = {'var_smoothing': np.logspace(0, -9, num=100)}
cv_strategy = KFold(n_splits=10, shuffle=True, random_state=42)

grid_search = GridSearchCV(
    estimator=gnb,
    param_grid=param_grid,
    cv=cv_strategy,
    scoring='f1_weighted',
    return_train_score=True,
    n_jobs=-1
)

grid_search.fit(X, y)

print("--- KẾT QUẢ TỐI ƯU THAM SỐ (GRIDSEARCHCV) ---")
print("Tham số var_smoothing tốt nhất:", grid_search.best_params_)
print("F1-score weighted cao nhất:", grid_search.best_score_)

# 3. Đánh giá 10-fold cross-validation
best_gnb = grid_search.best_estimator_
f1_scores_10fold = cross_val_score(best_gnb, X, y, cv=cv_strategy, scoring='f1_weighted')

print("\n--- CHI TIẾT 10-FOLD CROSS-VALIDATION ---")
for i, score in enumerate(f1_scores_10fold, 1):
    print(f"Fold {i}: F1-score = {score:.4f}")

print(f"\nF1-score Trung bình: {f1_scores_10fold.mean():.4f}")
print(f"Độ lệch chuẩn (Std): {f1_scores_10fold.std():.4f}")
```
